# NeoLM — Load Pretraining Corpus

Load the NeoCorpus foundation dataset for tokenizer training and model pretraining.

Dataset: [MaagDeveloper/NeoCorpus](https://huggingface.co/datasets/MaagDeveloper/NeoCorpus)

- Use the local dataset if available.
- Otherwise, download it from Hugging Face and save it locally.
- No corpus rebuilding or preprocessing is performed here.

The dataset contains `train`, `validation`, and `test` splits.

```python
from pathlib import Path
from datasets import load_dataset, load_from_disk

DATASET_ID = "MaagDeveloper/NeoCorpus"
CONFIG_NAME = "foundation"

DB_DIR = Path("../../artifacts/pretrain_db/final")

if DB_DIR.exists():
    pretrain_db = load_from_disk(str(DB_DIR))
    print("Loaded from local:", DB_DIR)
else:
    pretrain_db = load_dataset(DATASET_ID, CONFIG_NAME)

    DB_DIR.parent.mkdir(parents=True, exist_ok=True)
    pretrain_db.save_to_disk(str(DB_DIR))

    print("Downloaded from Hugging Face:", DATASET_ID)

print(pretrain_db)
```
